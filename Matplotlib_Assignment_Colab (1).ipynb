**Task 3: frequency vs density**

```python
# ---------- Task 3: frequency vs density ----------
steps = np.random.normal(8000, 2000, 30).astype(int).clip(2000, 15000)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))
a1.hist(steps, bins=8, edgecolor="black", color="seagreen")
a1.set_title("Frequency (default)")
a1.set_xlabel("Daily steps")
a1.set_ylabel("Count of days")
a2.hist(steps, bins=8, density=True, edgecolor="black", color="darkorange")
a2.set_title("density=True")
a2.set_xlabel("Daily steps")
a2.set_ylabel("Density")
fig.tight_layout()
plt.show()
# Explanation: default y-axis = raw COUNT of days per bin. With density=True the
# bar HEIGHTS are scaled so that the total AREA = 1 (height = count/(N*bin_width)),
# so the y-axis shows probability density, not counts. Shape is identical.
```

**Task 2 & 3: Zomato spend vs orders, labels + trend line**

```python
# ---------- Task 2 & 3: Zomato spend vs orders, labels + trend line ----------
names = ["Aarav", "Diya", "Rohan", "Isha", "Kabir", "Meera",
         "Vivaan", "Anaya", "Arjun", "Saanvi", "Reyansh", "Kiara"]
orders = np.array([4, 7, 10, 3, 12, 6, 15, 8, 5, 11, 9, 14])
spent = np.array([900, 1500, 2300, 700, 2700, 1400, 3400, 1800, 1100, 2500, 2000, 3100])

def zomato_plot(trend=False):
    fig, ax = plt.subplots(figsize=(9, 6))
    ax.scatter(orders, spent, color="crimson", zorder=3)
    for n, x, y in zip(names, orders, spent):
        ax.annotate(n, (x, y), textcoords="offset points", xytext=(6, 6), fontsize=9)
    if trend:
        m, c = np.polyfit(orders, spent, 1)
        xs = np.linspace(orders.min(), orders.max(), 100)
        ax.plot(xs, m * xs + c, "k--", label=f"Trend: y = {m:.1f}x + {c:.1f}")
        ax.legend()
    ax.set_title("Zomato: Orders vs Amount Spent")
    ax.set_xlabel("Number of orders placed")
    ax.set_ylabel("Amount spent (Rs.)")
    ax.grid(alpha=0.3)
    plt.show()

zomato_plot(trend=False)   # Task 2
zomato_plot(trend=True)    # Task 3
```

**Task 4: YouTube vs Spotify with equation on plot**

```python
# ---------- Task 4: YouTube vs Spotify with equation on plot ----------
np.random.seed(1)
yt = np.round(np.random.uniform(1, 8, 15), 1)
sp = np.round(0.6 * yt + np.random.normal(1, 0.6, 15), 1)
slope, intercept = np.polyfit(yt, sp, 1)
xs = np.linspace(yt.min(), yt.max(), 100)

fig, ax = plt.subplots()
ax.scatter(yt, sp, color="royalblue", label="Users")
ax.plot(xs, slope * xs + intercept, color="red", label="Regression line")
eq = f"y = {slope:.2f}x + {intercept:.2f}"
ax.text(0.05, 0.92, eq, transform=ax.transAxes, fontsize=12,
        bbox=dict(boxstyle="round", facecolor="wheat", alpha=0.8))
ax.set_title("YouTube Watch Hours vs Spotify Listening Hours")
ax.set_xlabel("YouTube watch hours")
ax.set_ylabel("Spotify listening hours")
ax.legend(loc="lower right")
print("Line equation:", eq)
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# ---------- Task 1: annotate the peak month ----------
dau = [3.8, 4.0, 4.2, 4.1, 4.5, 4.9, 5.2, 5.0, 5.4, 5.8, 5.6, 5.3]
peak = dau.index(max(dau))
plt.figure(figsize=(10, 5))
plt.plot(months, dau, marker="o")
plt.annotate(f"Peak: {months[peak]} ({dau[peak]}M)",
             xy=(peak, dau[peak]), xytext=(peak - 3, dau[peak] + 0.2),
             arrowprops=dict(arrowstyle="->", color="red"))
plt.ylim(3.5, 6.3)
plt.title("Music App: Daily Active Users (millions)")
plt.xlabel("Month")
plt.ylabel("DAU (millions)")
plt.show()
```

**Task 2: arrow at the lowest bar**

```python
# ---------- Task 2: arrow at the lowest bar ----------
orders = [320, 280, 350, 400, 380, 420, 390, 450, 430, 410, 460, 500]
low = orders.index(min(orders))
plt.figure(figsize=(10, 5))
plt.bar(months, orders, color="tomato")
plt.annotate("Lowest orders", xy=(low, orders[low]),
             xytext=(low + 1.5, orders[low] - 100 if orders[low] > 150 else 0),
             arrowprops=dict(facecolor="black", shrink=0.05, width=1.5))
plt.title("Zomato Monthly Orders (thousands)")
plt.xlabel("Month")
plt.ylabel("Orders (thousands)")
plt.show()
```

**Task 3: pie chart + legend at lower right**

```python
# ---------- Task 3: pie chart + legend at lower right ----------
methods = ["UPI", "Card", "Wallet", "Cash"]
share = [52, 25, 13, 10]
fig, ax = plt.subplots()
wedges, _, _ = ax.pie(share, autopct="%1.0f%%", startangle=90)
ax.legend(wedges, methods, title="Payment method", loc="lower right")
ax.set_title("Payment Methods on E-commerce App")
plt.show()
```

**Task 4: scatter + text box on highest-rated movie**

```python
# ---------- Task 4: scatter + text box on highest-rated movie ----------
movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E", "Movie F", "Movie G"]
rating = [7.2, 8.5, 6.4, 9.1, 7.8, 5.9, 8.0]
reviews = [1200, 3400, 800, 5200, 2100, 500, 2900]
top = rating.index(max(rating))
fig, ax = plt.subplots()
ax.scatter(reviews, rating, color="teal", s=70)
ax.annotate(f"Top rated: {movies[top]} ({rating[top]})",
            xy=(reviews[top], rating[top]), xytext=(reviews[top] - 3200, rating[top] - 0.9),
            bbox=dict(boxstyle="round,pad=0.4", fc="yellow", ec="black"),
            arrowprops=dict(arrowstyle="->"))
ax.set_title("Movie Ratings vs Number of Reviews")
ax.set_xlabel("Number of reviews")
ax.set_ylabel("Rating")
plt.show()
```

## Session 12 - Mini Case Study

**Task 1: create a sample dataset, load it, show head**

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

HERE = os.getcwd()
OUT = os.path.join(HERE, "report")
os.makedirs(OUT, exist_ok=True)

# ---------- Task 1: create a sample dataset, load it, show head ----------
csv_path = os.path.join(HERE, "sales_data.csv")
if not os.path.exists(csv_path):
    rng = np.random.default_rng(42)
    catalog = {"Electronics": [("Headphones", 2500), ("Smartwatch", 6000), ("Power Bank", 1500)],
               "Clothing": [("T-Shirt", 600), ("Jeans", 1800), ("Jacket", 3200)],
               "Groceries": [("Rice 5kg", 450), ("Snacks Pack", 250), ("Cooking Oil", 700)]}
    rows = []
    for _ in range(400):
        cat = rng.choice(list(catalog))
        prod, price = catalog[cat][rng.integers(0, 3)]
        qty = int(rng.integers(1, 4))
        date = pd.Timestamp("2025-01-01") + pd.Timedelta(days=int(rng.integers(0, 365)))
        rows.append((prod, cat, date.date(), price * qty))
    pd.DataFrame(rows, columns=["product", "category", "date", "sales"]).to_csv(csv_path, index=False)

df = pd.read_csv(csv_path, parse_dates=["date"])
print(df.head())
```

**Task 2 + 5: category bar chart with value labels, exported PNG**

```python
# ---------- Task 2 + 5: category bar chart with value labels, exported PNG ----------
cat_sales = df.groupby("category")["sales"].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(cat_sales.index, cat_sales.values, color=["#2E86AB", "#F6AE2D", "#33A02C"])
for b in bars:                                  # Task 5: value annotations
    ax.text(b.get_x() + b.get_width() / 2, b.get_height() + cat_sales.max() * 0.01,
            f"Rs. {b.get_height():,.0f}", ha="center", va="bottom", fontweight="bold")
ax.set_ylim(0, cat_sales.max() * 1.12)
ax.set_title("Total Sales by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Sales (Rs.)")
fig.tight_layout()
fig.savefig(os.path.join(OUT, "category_sales_kpi.png"), dpi=150)
plt.show()
```

**Task 3: monthly trend**

```python
# ---------- Task 3: monthly trend ----------
monthly = df.groupby(df["date"].dt.to_period("M"))["sales"].sum()
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(monthly.index.strftime("%b"), monthly.values, marker="o", color="teal", linewidth=2)
peak = monthly.values.argmax()
ax.annotate("Peak month", xy=(peak, monthly.values[peak]),
            xytext=(peak - 2, monthly.values[peak] * 1.05),
            arrowprops=dict(arrowstyle="->", color="red"))
ax.set_title("Monthly Sales Trend - 2025")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (Rs.)")
ax.grid(alpha=0.3)
fig.tight_layout()
fig.savefig(os.path.join(OUT, "monthly_trend.png"), dpi=150)
plt.show()
```

**Task 4: order-size histogram (bins = 10)**

```python
# ---------- Task 4: order-size histogram (bins = 10) ----------
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(df["sales"], bins=10, color="slateblue", edgecolor="black")
ax.set_title("Distribution of Order Amounts")
ax.set_xlabel("Order amount (Rs.)")
ax.set_ylabel("Number of orders")
fig.tight_layout()
fig.savefig(os.path.join(OUT, "order_distribution.png"), dpi=150)
plt.show()
```

**Extra KPI: top products**

```python
# ---------- Extra KPI: top products ----------
top = df.groupby("product")["sales"].sum().nlargest(5).sort_values()
fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(top.index, top.values, color="coral")
ax.set_title("Top 5 Products by Sales")
ax.set_xlabel("Sales (Rs.)")
fig.tight_layout()
fig.savefig(os.path.join(OUT, "top_products.png"), dpi=150)
plt.show()
```

**Export report (all KPI charts in one PDF)**

```python
# ---------- Export report (all KPI charts in one PDF) ----------
from matplotlib.backends.backend_pdf import PdfPages
from matplotlib.image import imread
with PdfPages(os.path.join(OUT, "sales_report.pdf")) as pdf:
    for name in ["category_sales_kpi", "monthly_trend", "order_distribution", "top_products"]:
        img = imread(os.path.join(OUT, f"{name}.png"))
        f = plt.figure(figsize=(8.27, 5.5))
        plt.imshow(img)
        plt.axis("off")
        pdf.savefig(f, bbox_inches="tight")
        plt.close(f)
print("Report saved to", OUT)
```
